```python
import numpy as np
import torch
```

```python
W = np.random.rand(4,5).astype(np.float32)
b = np.random.rand(1, 5).astype(np.float32)

x_1 = np.random.rand(1, 4).astype(np.float32)

pred_1 = x_1 @ W + b
print(pred_1, end="\n\n")

l_1 = np.sum(pred_1)
print(l_1)

dL = 1
dpred = dL * np.ones_like(pred_1)

db = np.sum(dpred, axis=0, keepdims=True)
dW = x_1.T @ dpred
dx = dpred @ W.T

print(dW, db)
```

```python
linear = torch.nn.Linear(4, 5)
linear.weight = torch.nn.Parameter(torch.tensor(W.T))
linear.bias = torch.nn.Parameter(torch.tensor(b, dtype=torch.float32))

x_2 = torch.tensor(x_1, dtype=torch.float32, requires_grad=True)

pred_2 = linear(x_2)
print(pred_2.detach().numpy(), end="\n\n")
l = torch.sum(pred_2)
l.backward()

print(linear.weight.grad)
```
