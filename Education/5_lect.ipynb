```python
from typing import Tuple, List, Optional
import numpy as np
```

```python
def creating_data(x_min, x_max, data_size, real_w, loc_n=0, scale_n=0, seed=42):
    
    rng = np.random.default_rng(seed)

    data = np.linspace(x_min, x_max, data_size, dtype=np.float32)
    noise = rng.normal(loc=loc_n, scale=scale_n, size=data.shape).astype(np.float32)

    x = data + noise
    xs = np.stack([np.ones_like(x), x], axis=1)
    ys = xs @ np.asarray(real_w, dtype=np.float32).reshape(2, 1)

    return (xs, ys)


def GD(xs, ys, init_value=0, alpha=0.1, n_iter=100):
    weight = np.full(shape=(xs.shape[1], ys.shape[1]), fill_value=init_value, dtype=np.float32)
    data_size = xs.shape[0]

    for epoch in range(n_iter):
        error = (xs @ weight - ys)
        loss = (error.T @ error) / data_size
        print(f"Epoch: {epoch + 1} -> loss = {loss}")

        grad = 2 * xs.T @ error /data_size
        weight -= grad * alpha

    return weight

def SGD(xs, ys, init_value=0, alpha=0.1, n_iter=100, batch_size=100, seed=42):
    weight = np.full(shape=(xs.shape[1], ys.shape[1]), fill_value=init_value, dtype=np.float32)
    data_size = xs.shape[0]
    count_batch = int(data_size/batch_size)
    rng = np.random.default_rng(seed)

    for epoch in range(n_iter):
        idx = rng.permutation(data_size)
        x_c = xs[idx]
        y_c = ys[idx]

        for i in range(count_batch):
            xb = x_c[batch_size * i: batch_size * (i + 1)]
            yb = y_c[batch_size * i: batch_size * (i + 1)]

            error = (xb @ weight - yb)
            loss = (error.T @ error) / data_size

            grad = 2 * xb.T @ error /data_size
            weight -= grad * alpha
        print()

    return weight
```
